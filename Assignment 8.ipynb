## Python Project: Online Shopping Cart

```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:
    print("\n===== ONLINE SHOPPING CART =====")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        print("\nAvailable Products")
        print("Product ID | Product | Price (Rs.) | Available Stock")
        for product_id, details in products.items():
            print(f"{product_id} | {details['name']} | {details['price']} | {details['stock']}")
            
    elif choice == "2":
        product_id = input("Enter Product ID (for example P1): ").strip().upper()

        if product_id in products:
            quantity = int(input("Enter quantity: "))

            if quantity <= 0:
                print("Please enter a quantity greater than zero.")
            else:
                if product_id in cart:
                    total_quantity = cart[product_id] + quantity
                else:
                    total_quantity = quantity

                if total_quantity <= products[product_id]["stock"]:
                    cart[product_id] = total_quantity
                    print("Product added to cart successfully.")
                else:
                    print("Not enough stock available.")
        else:
            print("Product ID not found. Please try again.")

    
    elif choice == "3":
        if not cart:
            print("Your cart is empty.")
        else:
            cart_total = 0
            print("\nYour Shopping Cart")
            print("Product | Quantity | Item Subtotal (Rs.)")

            for product_id, quantity in cart.items():
                details = products[product_id]
                subtotal = details["price"] * quantity
                cart_total += subtotal
                print(f"{details['name']} | {quantity} | {subtotal}")

            print("Cart Total: Rs.", cart_total)

    
    elif choice == "4":
        if not cart:
            print("Your cart is empty. Please add a product before checkout.")
        else:
            cart_total = 0

            for product_id, quantity in cart.items():
                details = products[product_id]
                subtotal = details["price"] * quantity
                cart_total += subtotal

            if cart_total >= 5000:
                discount = cart_total * 10 / 100
            else:
                discount = 0

            final_bill = cart_total - discount

            print("\n========== FINAL BILL ==========")
            print("Cart Total: Rs.", cart_total)
            print("Discount: Rs.", discount)
            print("Final Bill: Rs.", final_bill)

            
            for product_id, quantity in cart.items():
                products[product_id]["stock"] -= quantity

            
            cart.clear()
            print("Checkout successful. Stock updated and cart emptied.")

    
    elif choice == "5":
        print("Thank you for shopping. Goodbye!")
        break

    
    else:
        print("Invalid choice. Please select an option from 1 to 5.")
```
