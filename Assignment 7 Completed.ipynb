## Question 1 — Employee Data Bank using Class

```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("Employee Name:", self.name)
        print("Employee Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Years of Experience:", self.year_of_experience)

    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print(self.name, "'s experience has been updated.")

    def increase_salary(self, amount):
        self.salary = self.salary + amount
        print(self.name, "'s new salary is:", self.salary)
```

```python
employee101 = Employee(
    "Ali", 101, "Karachi", 75000, "Data Analyst",
    2022, "BS Computer Science", 4
)

print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)
print("Post:", employee101.post)
print("Year of Joining:", employee101.year_of_joining)
print("Education:", employee101.education)
print("Experience:", employee101.year_of_experience)

print("\nEmployee Details:")
employee101.describe_employee()

print("\nUpdating Experience:")
employee101.update_year_of_experience(5)

print("\nIncreasing Salary:")
employee101.increase_salary(10000)

print("\nUpdated Employee Details:")
employee101.describe_employee()
```

## Question 2 — Employee Data Bank without Classes or OOP

```python
# Function to display employee details
def describe_employee(employee):
    print("Employee Name:", employee["name"])
    print("Employee Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", employee["salary"])
    print("Post:", employee["post"])
    print("Year of Joining:", employee["year_of_joining"])
    print("Education:", employee["education"])
    print("Years of Experience:", employee["year_of_experience"])


# Function to update experience
def update_year_of_experience(employee, new_experience):
    employee["year_of_experience"] = new_experience
    print(employee["name"], "'s experience has been updated.")


# Function to increase salary
def increase_salary(employee, amount):
    employee["salary"] = employee["salary"] + amount
    print(employee["name"], "'s new salary is:", employee["salary"])
```

### Create employee101 without a class

Here, `employee101` is a dictionary, not an object of a class. We print its attributes individually and use the functions to display and update its information.
