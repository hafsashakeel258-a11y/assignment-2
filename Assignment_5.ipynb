## Q1. Count Vowels
Write a function that accepts a string and returns the number of vowels in it.

```python
def count_vowels(text):
    count = 0
    for letter in text:
        if letter in "aeiouAEIOU":
            count = count + 1
    return count

print(count_vowels("Welcome"))
```

## Q2. Display Student Information
Use keyword arguments for the student's details.

```python
def display_student_info(name, age, grade, school):
    print("Name:", name)
    print("Age:", age)
    print("Grade:", grade)
    print("School:", school)

display_student_info(
    name="John Doe",
    age=15,
    grade="10th",
    school="Greenwood High"
)
```

## Q3. Display Product Prices
Use `**kwargs` to accept any number of keyword arguments for prices.

```python
def display_prices(product, **prices):
    print("Product:", product)
    for store, price in prices.items():
        print()
        print(store + ": $" + str(price))

display_prices(
    "Laptop",
    Amazon=800,
    eBay=750,
    Walmart=780
)
```

## Q4. Find Maximum Number
Use `*args` to accept any number of positional arguments.

```python
def find_max(*numbers):
    return max(numbers)

print(find_max(10, 25, 15, 40, 8))
```

## Q5. Delivery Charges
Use a lambda function. Delivery costs Rs. 150 for orders below Rs. 2,000; otherwise delivery is free.

```python
delivery_charges = lambda order: 150 if order < 2000 else 0

print(delivery_charges(1500))
print(delivery_charges(2500))
```

## Q6. Rectangle Area and Perimeter
Use two nested functions and return both results.

```python
def calculate_rectangle(length, width):
    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()
    return area, perimeter

area, perimeter = calculate_rectangle(5, 3)
print("Area:", area)
print("Perimeter:", perimeter)
```

## Q8. Calculate Hotel Bill
Use a nested function. Room cost is Rs. 5,000 per night, plus 10% tax on room cost.

```python
def calculate_hotel_bill(nights):
    def calculate_room_cost():
        return nights * 5000

    def calculate_tax(room_cost):
        return room_cost * 10 / 100

    room_cost = calculate_room_cost()
    tax = calculate_tax(room_cost)
    total_bill = room_cost + tax

    print("Room Cost:", room_cost)
    print("Tax:", tax)
    print("Total Bill:", total_bill)

calculate_hotel_bill(3)
```

## Q9. User Login Validation
Use a nested function to check the username and password. This example uses `admin` and `12345` as the expected login details.

```python
def validate_login(username, password):
    def check_details():
        return username == "admin" and password == "12345"

    if check_details():
        return "Login successful"
    else:
        return "Invalid username or password"

print(validate_login("admin", "12345"))
print(validate_login("ali", "1111"))
```

## Q10. Shopping Bill with Discount
Use nested functions. The discount amount is subtracted from the total, but the final amount cannot be less than zero.

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price * quantity

    def apply_discount(total):
        final_amount = total - discount
        if final_amount < 0:
            final_amount = 0
        return final_amount

    total = calculate_total()
    return apply_discount(total)

print("Final Bill =", shopping_bill(300, 2), "PKR")
print("Final Bill =", shopping_bill(300, 2, 100), "PKR")
```
